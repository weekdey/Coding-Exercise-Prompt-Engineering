```python
def analyze_expenses(expenses):
    if not expenses:
        print("No expenses were provided.")
        return

    negative_expenses = [
        expense for expense in expenses
        if not isinstance(expense, (int, float))
        or isinstance(expense, bool)
        or expense < 0
    ]

    if negative_expenses:
        print(f"Invalid expense(s): {negative_expenses}")
        print("Expenses cannot be negative.")
        return

    total = sum(expenses)
    average = total / len(expenses)

    print(f"Total expense: ${total:.2f}")
    print(f"Average expense: ${average:.2f}")
```

```python
print("Test 1: Normal expenses")
analyze_expenses([25.50, 10.00, 42.75, 8.25])
```

```python
print("Test 2: Invalid expense")
analyze_expenses([25.50, -10.00, 42.75])
```

```python
print("Test 3: Empty list")
analyze_expenses([])
```
